```python
import pandas as pd
from sweetpea import Factor, CrossBlock, synthesize_trials, CMSGen

color = Factor("color", ["red", "blue"])
word = Factor("word", ["red", "blue"])

block = CrossBlock(design=[color, word], crossing=[color, word], constraints=[])
experiments = synthesize_trials(block, 1, CMSGen)

pd.DataFrame(experiments[0])
```
